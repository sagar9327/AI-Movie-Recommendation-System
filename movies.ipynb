```python
import pandas as pd

movies = pd.read_csv('tmdb_5000_movies.csv')
credits = pd.read_csv('tmdb_5000_credits.csv')
```

```python
data = movies.merge(credits, on='title')
```

```python
data['crew'][0]
```

```python
data = data[['genres','id','keywords','overview','release_date','title','cast','crew']]
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
data.dropna(inplace=True)
```

```python
data.isnull().sum()
```

```python
data.head(3)
```

```python
import ast

def genre_extract(col):
    l = []
    for i in ast.literal_eval(col):
        l.append(i['name'])
    return l

data['genres'] = data['genres'].apply(genre_extract)
```

```python
data.head(5)
```

```python
def keywords(col):
    l = []
    for i in ast.literal_eval(col):
        l.append(i['name'])
    return l

data['keywords'] = data['keywords'].apply(keywords)
```

```python
data.head(5)
```

```python
data['release_date'] = pd.to_datetime(data['release_date'])
```

```python
def cast_extract(col):
    l = []
    j = 1
    for i in ast.literal_eval(col):
        if j <= 3:
            l.append(i['name'])
            j += 1
    return l

```

```python
data['cast'] = data['cast'].apply(cast_extract)
```

```python
data.head(2)
```

```python
data['crew'][0]
```

```python
def crew_extract(col):
    l = []
    for i in ast.literal_eval(col):
        if i['job'] == 'Director':
            l.append(i['name'])
            break
    return l

data['crew'] = data['crew'].apply(crew_extract)
```

```python
data['overview'] = data['overview'].apply(lambda x:x.split())
```

```python
data['crew'] = data['crew'].apply(lambda x:[i.replace(" ", "") for i in x])
data['genres'] = data['genres'].apply(lambda x:[i.replace(" ", "") for i in x])
data['keywords'] = data['keywords'].apply(lambda x:[i.replace(" ", "") for i in x])
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
data['tags'][0]
```

```python
def stemming(text):
    l = []
    for i in text.split():
        l.append(ps.stem(i))
    return " ".join(l)

data['tags'] = data['tags'].apply(stemming)
```

```python
def recommend(movie):
    movie_index = data[data['title'] == movie].index[0]
    recommendations = similarities[movie_index]
    movie_list = sorted(
        enumerate(recommendations),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]
    for i in movie_list:
        print(data.iloc[i[0]].title)
    return recommendations
recommend('beowulf')
```
